```python
import torch
import numpy as np
```

```python
tensor_1 = torch.tensor([1,2,3,4])
tensor_2 = torch.tensor([5,6,7,8])
```

```python
#addiation Proccess
tensor_3 = tensor_1 + tensor_2
tensor_3
```

```python
#Addition Longhand
tensor_4 = torch.add(tensor_1, tensor_2)
tensor_4
```

```python
#another way
torch.add(tensor_1, tensor_2, out=tensor_3)
tensor_3
#or
tensor_3 = tensor_1.add(tensor_2)
tensor_3
```

```python
#Subtraction
tensor_5 = tensor_1 - tensor_2
tensor_5
```

```python
#Subtraction longhand
tensor_6 = torch.sub(tensor_1, tensor_2)
tensor_6
```

```python
#another way
torch.sub(tensor_1, tensor_2, out=tensor_5)
tensor_5
#or
tensor_5 = tensor_1.sub(tensor_2)
tensor_5
```

```python
#Multiplication
tensor_7 = tensor_1 * tensor_2
tensor_7
```

```python
#Multiplication Longhand
tensor_8 = torch.mul(tensor_1, tensor_2)
tensor_8
```

```python
#another way
torch.mul(tensor_1, tensor_2, out=tensor_7)
tensor_7
#or
tensor_7 = tensor_1.mul(tensor_2)
tensor_7
```

```python
#Divison
tensor_9 = tensor_2 / tensor_1
tensor_9
```

```python
#Divison Longhand
tensor_10 = torch.div(tensor_2, tensor_1)
tensor_10
```

```python
#another way
torch.div(tensor_2, tensor_1, out=tensor_9)
tensor_9
#or
tensor_9 = tensor_2.div(tensor_1)
tensor_9
```

```python
#remainder Modulus
tensor_11 = tensor_2 % tensor_1
tensor_11
```

```python
#remainder Modulus longhand
tensor_12 = torch.remainder(tensor_2, tensor_1)
tensor_12
```

```python
#another way
torch.remainder(tensor_2, tensor_1, out=tensor_11)
tensor_11
#or
tensor_11 = tensor_2.remainder(tensor_1)
tensor_11
```

```python
#Exponents / Powers
torch.pow(tensor_1, tensor_2)
```

```python
#another way
tensor_13 = tensor_1.pow(tensor_2)
tensor_13
```
